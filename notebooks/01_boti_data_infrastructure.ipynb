## 4. Test Asynchronous Model Builders
Runs natively inside IPython's loop instantiating native async connections wrapping declarative boundaries safely.

```python
import asyncio
from boti_data.db.sql_model_builder import SqlAlchemyModelBuilder, BuilderConfig

async def test_async_buildings():
    # Overriding standard Module namespaces securely via Immutable Pydantic Configs
    b_config = BuilderConfig(module_label="boti_data.analytics.enterprise")
    
    async with AsyncSqlDatabaseResource(SqlDatabaseConfig(connection_url=SecretStr(async_db_dsn))) as async_db:
        builder = SqlAlchemyModelBuilder(async_db.engine, "asm_tracking_productos", config=b_config)
        
        # Execute the awaitable sequence spanning Async engine threads seamlessly
        AsyncUserModel = await builder.build_model_async()
        
        print(f"Reflected Async Bound Class: {AsyncUserModel}")
        print(f"Configured Namespace Target Tracking: {AsyncUserModel.__module__}")

        # Due to memory caching registries natively preserving the structural engine constraints, 
        # this safely operates across the exact same structural bounds!
        return AsyncUserModel

# Trigger explicit evaluation safely using native loop tracks
await test_async_buildings()
```
