# Boti notebook test setup

Use this notebook first to confirm the repository path and module imports are working before running the focused module notebooks.

```python
import importlib
import sys
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "src" / "boti").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent.resolve()

SRC_ROOT = PROJECT_ROOT / "src"
if str(SRC_ROOT) not in sys.path:
    sys.path.insert(0, str(SRC_ROOT))

PROJECT_ROOT
```

```python
modules = [
    "boti",
    "boti.main",
    "boti.core",
    "boti.core.models",
    "boti.core.project",
    "boti.core.security",
    "boti.core.logger",
    "boti.core.managed_resource",
    "boti.core.secure_io",
]

loaded = {name: importlib.import_module(name).__name__ for name in modules}
loaded
```

```python
from boti.core import Logger, ManagedResource, ProjectService, SecureResource, is_secure_path

assert callable(Logger.default_logger)
assert ManagedResource.__name__ == "ManagedResource"
assert ProjectService.__name__ == "ProjectService"
assert SecureResource.__name__ == "SecureResource"
assert callable(is_secure_path)

print("Notebook imports are ready.")
```
