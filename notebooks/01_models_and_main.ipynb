# Models and CLI smoke tests

This notebook exercises `boti.core.models` and the CLI entry point in `boti.main`.

```python
import io
import sys
from contextlib import redirect_stdout
from pathlib import Path

PROJECT_ROOT = Path.cwd().resolve()
if not (PROJECT_ROOT / "src" / "boti").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent.resolve()

SRC_ROOT = PROJECT_ROOT / "src"
if str(SRC_ROOT) not in sys.path:
    sys.path.insert(0, str(SRC_ROOT))
```

```python
from boti.core.models import LoggerConfig, ResourceConfig
from boti.main import main

logger_config = LoggerConfig(logger_name="notebook-models", log_dir=PROJECT_ROOT / "logs")
assert logger_config.log_dir.is_absolute()
assert logger_config.log_file is None

resource_a = ResourceConfig(project_root=PROJECT_ROOT)
resource_b = ResourceConfig(project_root=PROJECT_ROOT)
resource_a.extra_allowed_paths.append(PROJECT_ROOT / "tmp")
assert resource_b.extra_allowed_paths == []

print(logger_config)
print(resource_a)
```

```python
buffer = io.StringIO()
with redirect_stdout(buffer):
    main()

output = buffer.getvalue()
assert "Boti Initialized." in output
assert "Python version:" in output

print(output)
```
