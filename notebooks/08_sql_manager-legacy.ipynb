```python
with SqlDatabaseResource(sync_config) as db:
    products_model = SqlAlchemyModelBuilder(db.engine, "asm_tracking_productos").build_model()

    stmt = (
        select(
            products_model.id_track_global,
            products_model.id_producto,
            products_model.codigo_barra,
        )
        .where(
            products_model.id_tipo_producto == 1,
            products_model.id_track_global.in_([1, 2, 3, 4]),
        )
    )

    with db.session() as session:
        active_products = session.execute(stmt).all()
```
